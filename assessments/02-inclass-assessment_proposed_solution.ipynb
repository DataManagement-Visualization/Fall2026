### Part B — Match the connections
Each tuple is `(player_1, player_2, date)`. The two lists record when
players became friends and when they unfriended. `(100, 101)` and
`(101, 100)` are the same pair.

Complete `merge_connections(friendships, unfriendships)` to **return a list**:

- One tuple per friendship: `(smaller_ID, larger_ID, start_date, end_date)`.
- Use the matching unfriendship date, or `"ongoing"` if there is no match.
- Keep the friendship list's order; ignore unfriendships with no friendship.
- Leave both input lists unchanged.

Assume valid inputs, distinct IDs, and at most one record per pair in
each list. Dates are strings; do not parse them. Use loops, conditionals,
and lists/tuples. No imports are needed.

````python
friendships = [
    (100, 101, "09/01/2026"),
    (105, 102, "09/02/2026"),
    (110, 111, "09/03/2026")
]
unfriendships = [
    (101, 100, "09/08/2026"),
    (102, 105, "09/10/2026"),
    (120, 121, "09/11/2026")
]
````

**Expected output for these inputs**

```python
[
    (100, 101, "09/01/2026", "09/08/2026"),
    (102, 105, "09/02/2026", "09/10/2026"),
    (110, 111, "09/03/2026", "ongoing")
]
```

````python
# Match each friendship with its end date, if one exists.
def merge_connections(friendships, unfriendships):
    merged = []
    for friendship in friendships:
        p1 = friendship[0]
        p2 = friendship[1]
        if p1 < p2:
            smaller = p1
            larger = p2
        else:
            smaller = p2
            larger = p1

        end_date = "ongoing"
        for unfriendship in unfriendships:
            q1 = unfriendship[0]
            q2 = unfriendship[1]
            if (p1 == q1 and p2 == q2) or (p1 == q2 and p2 == q1):
                end_date = unfriendship[2]

        merged.append((smaller, larger, friendship[2], end_date))
    return merged
````
