## 3. Prepare the assessment directory

```bash
mkdir econ525_assessment
cp examples/nursery.txt econ525_assessment/sample.txt
cd econ525_assessment
pwd
nano my_script.sh
```


## 4. Contents of `my_script.sh`

```bash
#!/usr/bin/env bash

echo "Hello, Data Science!"

cat sample.txt
head -n 7 sample.txt
tail -n 7 sample.txt

ls -lh > directory_list.txt
sort directory_list.txt | less

mkdir texts
: > texts/all_texts.txt

for i in {1..5}
do
  echo "This is file $i." > "texts/file$i.txt"
  cat "texts/file$i.txt" >> texts/all_texts.txt
done
```

The command `: > texts/all_texts.txt` creates an empty output file or clears an existing one. The loop appends only `file1.txt` through `file5.txt`, so the output file never reads itself.


## 5. Run and verify

```bash
bash my_script.sh
ls -lh
ls -lh texts
cat texts/all_texts.txt
```

Expected final text:

```text
This is file 1.
This is file 2.
This is file 3.
This is file 4.
This is file 5.
```

This script is intended to run once in a newly created `econ525_assessment` directory. If you run it again, `mkdir texts` will report that the directory already exists.
