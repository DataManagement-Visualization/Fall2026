**Pseudocode for the wheel-count rule**

```text
object = camera.get_object()
if object.has_wheels(): # does the object have wheels?
    if len(object.wheels) == 4: return "Car" # four wheels => car    
    elif len(object.wheels) == 2:
        if object.seen_from_back():

```

### What does one handwritten digit look like as data?

Each of the 1,797 handwriting samples has a known label from 0 through 9. We can look up the image, its features, and its label using the same index `i`.

| In Python | What you get for sample `i` |
|---|---|
| `digits.images[i]` | An image with 8 rows and 8 columns. Each pixel has a value from 0 (no ink) to 16 (most ink). |
| `digits.data[i]` | One table row with 64 features, one per pixel position. Column 0 is the top-left pixel, column 7 is the top-right, and column 8 starts the next image row. |
| `digits.target[i]` | The digit the person wrote. This label tells us what we want the model to predict. |

`.shape` gives the size of each dimension. The first number in each output counts handwriting samples. [Digits dataset](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_digits.html)

````python
digits = datasets.load_digits()
print("Image grids:", digits.images.shape)
print("Feature table:", digits.data.shape)
print("Labels:", digits.target.shape)
````

`plt.subplots(1, 4)` gives us four panels in one row. `zip` pairs each panel with one of the first four images and its known label. `imshow` draws the 8 × 8 grid. The reversed grayscale (`gray_r`) puts dark ink on a light background, and `nearest` keeps the edges of each pixel visible.

````python
# Adapted from https://scikit-learn.org/stable/auto_examples/classification/plot_digits_classification.html
_, axes = plt.subplots(1, 4)
for ax, image, label in zip(axes, digits.images[:4], digits.target[:4]):
    ax.imshow(image, cmap="gray_r", vmin=0, vmax=16, interpolation="nearest")
    ax.set_title(f"Label: {label}")
    ax.set_axis_off()
plt.show()
````

### From an image to one row of features

The classifier expects one row per example. **Flattening** reads the pixels across the first image row, then across the second, and continues to the bottom. For example, two rows `[0, 3]` and `[8, 1]` become one row `[0, 3, 8, 1]`.

For our 8 × 8 images, `reshape` keeps all 1,797 samples separate. The `-1` tells NumPy to work out the row length, which is 64. Scikit-learn already provides this layout in `digits.data`. Compare the top row of the first image with the first eight entries of its flattened row below.

````python
X = digits.images.reshape(len(digits.images), -1)
print("Top row of the first image:", digits.images[0, 0])
print("First eight features:      ", X[0, :8])
````

### How can a neural network recognize a digit?

A single dark pixel tells us little. Several dark pixels in a row might form the top stroke of a 7. A **neural network** learns which combinations of pixels help it recognize each digit. Its **hidden layer** combines the 64 inputs using adjustable numbers called *weights*. Its output layer uses those combinations to score the ten possible digits.

During training, the network compares its scores with the known labels and adjusts its weights to give the correct digits higher scores. For a new image, it predicts the digit with the highest score. This is one example of a neural network. We'll look more closely at how the weights are learned later.

In Python, `MLPClassifier` creates this type of network (a *multilayer perceptron*). `fit(X_train, y_train)` learns from images paired with their labels. `predict(X_test)` takes only the pixels for images held out of training.

````python
# Keep the first half for training and the remaining images for testing.
X_train, X_test, y_train, y_test = train_test_split(
    X, digits.target, test_size=0.5, shuffle=False
)

# alpha sets the penalty on large weights. random_state makes training repeatable.
classifier = MLPClassifier(alpha=0.001, random_state=0)
classifier.fit(X_train, y_train)
predicted = classifier.predict(X_test)
````

Now look at the first four test images. `reshape(8, 8)` puts each row of 64 features back into a grid for plotting. Each title pairs the prediction with the recorded label from `y_test`. For example, `Prediction: 6 | Label: 8` means the network read an 8 as a 6.

````python
_, axes = plt.subplots(1, 4)
for ax, row, prediction, label in zip(axes, X_test[:4], predicted[:4], y_test[:4]):
    ax.imshow(row.reshape(8, 8), cmap="gray_r", vmin=0, vmax=16, interpolation="nearest")
    ax.set_title(f"Prediction: {prediction} | Label: {label}")
    ax.set_axis_off()
plt.show()
````

# Recall: Supervised Learning

The most common approach to machine learning is supervised learning.

<center><img width=70% src="img/tesla_data.png"/></center>

1. First, we collect a dataset of labeled training examples.
2. We train a model to output accurate predictions on this dataset.
3. When the model sees new, similar data, it will also be accurate.

Let’s start by dissecting a simple example of a supervised learning problem.

# A Recipe for Applying Supervised Learning

To apply supervised learning, we define a dataset and a learning algorithm.

$$ \text{Dataset} + \text{Learning Algorithm} \to \text{Predictive Model} $$

The output is a predictive model that maps inputs to targets. For instance, it can predict targets on new inputs.

# A Supervised Learning Dataset

Can baseline body mass index (BMI) help us predict diabetes progression one year later? The dataset contains 442 patients and 10 baseline measurements per patient. For this example, our **input** is one column, BMI. Our **target** is each patient's recorded progression score one year later. [Diabetes dataset](https://scikit-learn.org/stable/modules/generated/sklearn.datasets.load_diabetes.html)

**Why multiply by 30 and add 25?** The loader centers and scales BMI, so zero represents the average BMI across all 442 patients. We choose 30 to spread out the small differences and 25 to place the average at a familiar-looking value. A scaled BMI of 0 becomes 25, and 0.05 becomes `30 × 0.05 + 25 = 26.5`. The patients keep their order on the horizontal axis, and their progression scores stay the same.

These numbers are on an illustrative scale. To get the original BMI measurements in kg/m², use `datasets.load_diabetes(as_frame=True, scaled=False).data["bmi"]`.

````python
# Return inputs and target separately as a pandas DataFrame and Series.
diabetes_X, diabetes_y = datasets.load_diabetes(return_X_y=True, as_frame=True)
diabetes_X = diabetes_X.loc[:, ['bmi']]
diabetes_X = diabetes_X * 30 + 25

# Use the last 20 patients for training and display the first five of those.
diabetes_X_train = diabetes_X.iloc[-20:]
diabetes_y_train = diabetes_y.iloc[-20:]
pd.concat([diabetes_X_train, diabetes_y_train], axis=1).rename(
    columns={'bmi': 'BMI (illustrative scale)', 'target': 'Progression score'}
).head().round(2)
````

````python
plt.scatter(diabetes_X_train, diabetes_y_train)
plt.xlabel('BMI (illustrative scale)')
plt.ylabel('One-year progression score')
plt.title('BMI and disease progression for the 20 training patients')
plt.show()
````

# The Model Family

Let's assume that the progression score is a linear function of BMI. In other words, for some unknown $\theta_0, \theta_1 \in \mathbb{R}$, we have
$$ y = \theta_1 \cdot x + \theta_0, $$
where $x$ is BMI on our illustrative scale (also called the independent variable), and $y$ is the one-year progression score (the dependent variable). 

The parameters $\theta_1, \theta_0$ are the slope and the intercept of the line that relates $x$ to $y$.

````python
theta_list = [(1, 2), (2,1), (1,0), (0,1)]
for theta0, theta1 in theta_list:
    x = np.arange(10)
    y = theta1 * x + theta0
    plt.plot(x,y, label=f'intercept={theta0}, slope={theta1}')
plt.xlabel('$x$ (illustrative BMI scale)')
plt.ylabel('$y$ (progression score)')
plt.legend()
plt.show()
````

# A Supervised Learning Algorithm: The Optimizer

Given our assumption that $x,y$ follow a linear relationship, the goal of a supervised learning algorithm is to find a good set of parameters consistent with the data.

This is an optimization problem.
For now, let's call the `sklearn.linear_model` library to find a $\theta_1, \theta_0$ that fit the data well.

````python
# Create linear regression object
regr = linear_model.LinearRegression()

# Train the model using the training sets
regr.fit(diabetes_X_train, diabetes_y_train.values)

# Make predictions on the training set
diabetes_y_train_pred_bmi_only = regr.predict(diabetes_X_train)

# The coefficients
print(f'Slope (theta1): {regr.coef_[0]:.2f}')
print(f'Intercept (theta0): {regr.intercept_:.2f}')
````

````python
plt.xlabel('BMI (illustrative scale)')
plt.ylabel('One-year progression score')
plt.scatter(diabetes_X_train, diabetes_y_train, label='Data')
label = fr'Model --- slope ($\theta_1$) is ${regr.coef_[0]:.2f}$, intercept ($\theta_0$) is ${regr.intercept_:.2f}$'
plt.plot(diabetes_X_train, diabetes_y_train_pred_bmi_only, color='black', linewidth=2,
         label=label)
plt.legend();
````

# Making New Predictions

For a new patient, we put BMI on the same scale used for training and use the fitted line to predict the one-year progression score.

Given an $x_\text{new}$, we can output prediction $y_\text{new}$ as
$$ y_\text{new} = f(x_\text{new}) = \theta_1^* \cdot x_\text{new} + \theta_0. $$

````python
# Compare the training patients with three patients held out for testing.
test_indices = [0, 4, 7]
diabetes_X_test = diabetes_X.iloc[test_indices]
diabetes_y_test = diabetes_y.iloc[test_indices]

plt.scatter(diabetes_X_train, diabetes_y_train, label='Initial patients (training data)')
plt.scatter(diabetes_X_test, diabetes_y_test, color='red', label='New patients (test data)')
plt.xlabel('BMI (illustrative scale)')
plt.ylabel('One-year progression score')
plt.legend()
plt.show()
````

The fitted line gives one predicted progression score for each new patient. In the plot below, the red circle shows the recorded score and the red cross shows the prediction at the same BMI.

````python
# generate predictions on the new patients
diabetes_y_test_pred_bmi_only = regr.predict(diabetes_X_test)

# visualize the results
plt.xlabel('BMI (illustrative scale)')
plt.ylabel('One-year progression score')
plt.scatter(diabetes_X_train, diabetes_y_train)
plt.scatter(diabetes_X_test, diabetes_y_test, color='red', marker='o')
plt.plot(diabetes_X_train, diabetes_y_train_pred_bmi_only, color='black', linewidth=1)
plt.plot(diabetes_X_test, diabetes_y_test_pred_bmi_only, 'x', color='red', mew=3, markersize=8)
plt.legend(['Initial patients (training data)', 'New patients (test data)',
            "Trained linear model", "Model predictions on test data"]);
````
